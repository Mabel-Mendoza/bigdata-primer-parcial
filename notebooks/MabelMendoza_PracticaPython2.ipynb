## Ejercicio 16: DataFrame de estudiantes
**Enunciado:** Crear un DataFrame con al menos 10 estudiantes que incluya nombre, carrera, edad, promedio y ciudad. Mostrar las primeras 5 filas, la información general, el promedio general, los estudiantes con promedio mayor o igual a 80 y ordenarlos de mayor a menor promedio.

```python
# Importamos pandas para poder usar DataFrames
import pandas as pd

# Creamos los datos de los estudiantes en un diccionario
datos = {
    "nombre":   ["Ana López", "Carlos Pérez", "María García", "José Martínez", "Laura Díaz",
                 "Pedro Ramírez", "Sofía Torres", "Luis Hernández", "Valeria Cruz", "Diego Flores"],
    "carrera":  ["Sistemas", "Contabilidad", "Derecho", "Medicina", "Sistemas",
                 "Derecho", "Medicina", "Contabilidad", "Sistemas", "Medicina"],
    "edad":     [20, 22, 21, 23, 20, 24, 21, 22, 20, 23],
    "promedio": [85, 72, 90, 65, 88, 78, 95, 70, 83, 60],
    "ciudad":   ["Tegucigalpa", "San Pedro Sula", "Tegucigalpa", "La Ceiba", "Comayagua",
                 "San Pedro Sula", "Tegucigalpa", "La Ceiba", "Comayagua", "Tegucigalpa"]
}

# Convertimos el diccionario en un DataFrame
df = pd.DataFrame(datos)

# Mostramos las primeras 5 filas
print("Primeras 5 filas:")
print(df.head())

# Mostramos la información general del DataFrame
print("\nInformación general:")
print(df.info())

# Calculamos el promedio general de todos los estudiantes
promedio_general = df["promedio"].mean()
print("\nPromedio general de calificaciones:", promedio_general)

# Mostramos solo los estudiantes con promedio mayor o igual a 80
print("\nEstudiantes con promedio mayor o igual a 80:")
print(df[df["promedio"] >= 80])

# Ordenamos de mayor a menor promedio
print("\nEstudiantes ordenados por promedio (mayor a menor):")
print(df.sort_values("promedio", ascending=False))
```

**Explicación de métodos pandas:**
- `pd.DataFrame()` convierte un diccionario en una tabla organizada en filas y columnas.
- `head()` muestra solo las primeras 5 filas, para revisar rápido cómo quedaron los datos.
- `info()` muestra un resumen del DataFrame: cuántas filas hay, qué tipo de dato tiene cada columna y si hay valores vacíos.
- `mean()` calcula el promedio de todos los valores de una columna.
- `df[df["promedio"] >= 80]` filtra la tabla y muestra solo las filas que cumplen esa condición.
- `sort_values()` ordena la tabla según una columna. Con `ascending=False` ordena de mayor a menor.

## Ejercicio 17: Análisis de ventas con pandas
**Enunciado:** Crear un DataFrame de ventas con al menos 15 registros que incluya fecha, vendedor, producto, categoría, cantidad y precio unitario. Calcular una columna de total por venta, el total por vendedor, el producto más vendido, la categoría con mayor ingreso y ordenar las ventas de mayor a menor.

```python
import pandas as pd

# Creamos los datos de ventas
datos_ventas = {
    "fecha":          ["2024-01-05", "2024-01-06", "2024-01-07", "2024-01-08", "2024-01-09",
                       "2024-01-10", "2024-01-11", "2024-01-12", "2024-01-13", "2024-01-14",
                       "2024-01-15", "2024-01-16", "2024-01-17", "2024-01-18", "2024-01-19"],
    "vendedor":       ["Ana", "Carlos", "Ana", "Pedro", "Carlos",
                       "Ana", "Pedro", "Carlos", "Ana", "Pedro",
                       "Carlos", "Ana", "Pedro", "Carlos", "Ana"],
    "producto":       ["Laptop", "Mouse", "Teclado", "Laptop", "Monitor",
                       "Mouse", "Teclado", "Laptop", "Monitor", "Mouse",
                       "Teclado", "Laptop", "Monitor", "Mouse", "Teclado"],
    "categoria":      ["Computación", "Accesorios", "Accesorios", "Computación", "Computación",
                       "Accesorios", "Accesorios", "Computación", "Computación", "Accesorios",
                       "Accesorios", "Computación", "Computación", "Accesorios", "Accesorios"],
    "cantidad":       [2, 5, 3, 1, 2, 8, 4, 3, 1, 6, 5, 2, 3, 7, 4],
    "precio_unitario":[15000, 250, 500, 15000, 5000, 250, 500, 15000, 5000, 250, 500, 15000, 5000, 250, 500]
}

# Convertimos a DataFrame
df_ventas = pd.DataFrame(datos_ventas)

# Calculamos el total por venta
df_ventas["TotalVenta"] = df_ventas["cantidad"] * df_ventas["precio_unitario"]

# Total vendido por vendedor
print("Total vendido por vendedor:")
print(df_ventas.groupby("vendedor")["TotalVenta"].sum())

# Producto más vendido por cantidad
print("\nProducto más vendido:")
print(df_ventas.groupby("producto")["cantidad"].sum().idxmax())

# Categoría con mayor ingreso
print("\nCategoría con mayor ingreso:")
print(df_ventas.groupby("categoria")["TotalVenta"].sum().idxmax())

# Ventas ordenadas de mayor a menor
print("\nVentas ordenadas de mayor a menor:")
print(df_ventas.sort_values("TotalVenta", ascending=False))
```

**Explicación de métodos pandas:**
- `df["TotalVenta"] = df["cantidad"] * df["precio_unitario"]` crea una columna nueva multiplicando dos columnas existentes.
- `groupby()` agrupa los datos por una columna, por ejemplo todos los registros del mismo vendedor juntos.
- `sum()` suma todos los valores del grupo.
- `idxmax()` devuelve el nombre del grupo que tiene el valor más alto.
- `sort_values()` ordena la tabla por la columna indicada.

## Ejercicio 18: Limpieza de datos con pandas
**Enunciado:** Crear un DataFrame con datos incompletos de clientes que incluya nombre, edad, correo, ciudad y teléfono. Identificar valores nulos, contarlos por columna, reemplazar ciudades vacías, eliminar registros sin correo y mostrar el DataFrame limpio.

```python
import pandas as pd
import numpy as np

# Creamos el DataFrame con datos incompletos usando None para simular vacíos
datos_clientes = {
    "nombre":    ["Ana López", "Carlos Pérez", "María García", "José Martínez", "Laura Díaz",
                  "Pedro Ramírez", "Sofía Torres"],
    "edad":      [25, None, 30, 22, None, 35, 28],
    "correo":    ["ana@mail.com", None, "maria@mail.com", "jose@mail.com", "laura@mail.com",
                  None, "sofia@mail.com"],
    "ciudad":    ["Tegucigalpa", "San Pedro Sula", None, "La Ceiba", None,
                  "Tegucigalpa", "Comayagua"],
    "telefono":  ["9999-0001", "9999-0002", None, "9999-0004", "9999-0005",
                  None, "9999-0007"]
}

df_clientes = pd.DataFrame(datos_clientes)

# Mostramos el DataFrame original
print("DataFrame original:")
print(df_clientes)

# Identificamos los valores nulos
print("\nValores nulos en el DataFrame:")
print(df_clientes.isnull())

# Contamos los valores nulos por columna
print("\nConteo de valores nulos por columna:")
print(df_clientes.isnull().sum())

# Reemplazamos ciudades vacías con el texto No especificada
df_clientes["ciudad"] = df_clientes["ciudad"].fillna("No especificada")

# Eliminamos los registros que no tienen correo
df_clientes = df_clientes.dropna(subset=["correo"])

# Mostramos el DataFrame limpio
print("\nDataFrame limpio:")
print(df_clientes)
```

**Explicación de métodos pandas:**
- `isnull()` revisa cada celda del DataFrame y muestra True si está vacía o False si tiene datos.
- `isnull().sum()` cuenta cuántos valores vacíos hay en cada columna.
- `fillna()` reemplaza los valores vacíos de una columna con el texto que se le indique.
- `dropna(subset=["correo"])` elimina las filas completas donde la columna correo esté vacía.

## Ejercicio 19: Análisis de préstamos
**Enunciado:** Crear un DataFrame con solicitudes de préstamo que incluya cliente, monto solicitado, plazo, ingreso mensual y estado de solicitud. Calcular el monto promedio, contar solicitudes por estado, mostrar solo los aprobados, los que superan 50,000 y agrupar por estado calculando el promedio del monto.

```python
import pandas as pd

# Creamos los datos de solicitudes de préstamo
datos_prestamos = {
    "cliente":          ["Ana López", "Carlos Pérez", "María García", "José Martínez", "Laura Díaz",
                         "Pedro Ramírez", "Sofía Torres", "Luis Hernández", "Valeria Cruz", "Diego Flores"],
    "monto_solicitado": [30000, 75000, 50000, 120000, 45000, 80000, 25000, 60000, 90000, 35000],
    "plazo_meses":      [12, 24, 18, 36, 12, 24, 6, 18, 36, 12],
    "ingreso_mensual":  [8000, 15000, 12000, 25000, 9000, 18000, 6000, 13000, 20000, 7000],
    "estado":           ["Aprobado", "Rechazado", "Aprobado", "Aprobado", "Pendiente",
                         "Aprobado", "Rechazado", "Pendiente", "Aprobado", "Rechazado"]
}

df_prestamos = pd.DataFrame(datos_prestamos)

# Calculamos el monto promedio solicitado
promedio_monto = df_prestamos["monto_solicitado"].mean()
print("Monto promedio solicitado:", promedio_monto)

# Contamos solicitudes por estado
print("\nCantidad de solicitudes por estado:")
print(df_prestamos["estado"].value_counts())

# Mostramos solo los préstamos aprobados
print("\nPréstamos aprobados:")
print(df_prestamos[df_prestamos["estado"] == "Aprobado"])

# Mostramos clientes con monto mayor a 50000
print("\nClientes con monto solicitado mayor a 50,000:")
print(df_prestamos[df_prestamos["monto_solicitado"] > 50000])

# Agrupamos por estado y calculamos el promedio del monto
print("\nPromedio de monto solicitado por estado:")
print(df_prestamos.groupby("estado")["monto_solicitado"].mean())
```

**Explicación de métodos pandas:**
- `mean()` calcula el promedio de todos los valores de una columna.
- `value_counts()` cuenta cuántas veces aparece cada valor diferente en una columna, en este caso cuántas solicitudes hay por cada estado.
- `df[df["estado"] == "Aprobado"]` filtra la tabla y muestra solo las filas donde el estado sea exactamente ese valor.
- `groupby()` agrupa los registros por categoría y permite hacer cálculos por grupo, como el promedio de cada estado.

## Ejercicio 20: Mini análisis de datos tipo empresa
**Enunciado:** Crear un DataFrame con información de empleados que incluya nombre, departamento, cargo, salario, años de experiencia y ciudad. Mostrar los primeros registros, estadísticas de salarios, salario promedio por departamento, el empleado con salario más alto, empleados con más de 5 años de experiencia, ordenar por salario y crear una columna de nivel de experiencia.

```python
import pandas as pd

# Creamos los datos de empleados
datos_empleados = {
    "nombre":      ["Ana López", "Carlos Pérez", "María García", "José Martínez", "Laura Díaz",
                    "Pedro Ramírez", "Sofía Torres", "Luis Hernández", "Valeria Cruz", "Diego Flores"],
    "departamento":["Sistemas", "Ventas", "Recursos Humanos", "Sistemas", "Ventas",
                    "Contabilidad", "Sistemas", "Ventas", "Contabilidad", "Recursos Humanos"],
    "cargo":       ["Analista", "Vendedor", "Coordinadora", "Desarrollador", "Vendedora",
                    "Contador", "Desarrolladora", "Supervisor", "Asistente", "Asistente"],
    "salario":     [25000, 18000, 22000, 30000, 17000, 28000, 32000, 20000, 15000, 14000],
    "experiencia": [3, 1, 6, 8, 2, 7, 9, 4, 1, 2],
    "ciudad":      ["Tegucigalpa", "San Pedro Sula", "Tegucigalpa", "La Ceiba", "Comayagua",
                    "San Pedro Sula", "Tegucigalpa", "La Ceiba", "Comayagua", "Tegucigalpa"]
}

df_empleados = pd.DataFrame(datos_empleados)

# Mostramos los primeros registros
print("Primeros registros:")
print(df_empleados.head())

# Estadísticas generales de salarios
print("\nEstadísticas generales de salarios:")
print(df_empleados["salario"].describe())

# Salario promedio por departamento
print("\nSalario promedio por departamento:")
print(df_empleados.groupby("departamento")["salario"].mean())

# Empleado con salario más alto
print("\nEmpleado con salario más alto:")
print(df_empleados[df_empleados["salario"] == df_empleados["salario"].max()])

# Empleados con más de 5 años de experiencia
print("\nEmpleados con más de 5 años de experiencia:")
print(df_empleados[df_empleados["experiencia"] > 5])

# Ordenamos por salario de mayor a menor
print("\nEmpleados ordenados por salario:")
print(df_empleados.sort_values("salario", ascending=False))

# Creamos la columna NivelExperiencia
def nivel_experiencia(anios):
    if anios <= 2:
        return "Junior"
    elif anios <= 5:
        return "Intermedio"
    else:
        return "Senior"

df_empleados["NivelExperiencia"] = df_empleados["experiencia"].apply(nivel_experiencia)

print("\nDataFrame con columna NivelExperiencia:")
print(df_empleados[["nombre", "experiencia", "NivelExperiencia"]])
```

**Explicación de métodos pandas:**
- `describe()` muestra un resumen estadístico de una columna numérica: el mínimo, máximo, promedio, y otros valores útiles de un solo vistazo.
- `groupby()` con `mean()` agrupa por departamento y calcula el salario promedio de cada uno.
- `max()` devuelve el valor más alto de una columna, y con el filtro se muestra la fila completa de ese empleado.
- `apply()` aplica una función a cada valor de una columna. En este caso se usó para asignar el nivel de experiencia según los años, creando una columna nueva con el resultado.
