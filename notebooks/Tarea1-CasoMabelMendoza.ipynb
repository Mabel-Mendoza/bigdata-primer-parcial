## Propósito de la actividad

En esta actividad se analiza cómo Spotify, una de las plataformas de música en streaming más grandes del mundo, utiliza Big Data para entender el comportamiento de sus usuarios y ofrecerles una experiencia musical completamente personalizada. A través de este caso se busca conectar los conceptos teóricos vistos en clase con una aplicación real y cotidiana, demostrando que el Big Data es algo que millones de personas experimentan cada vez que abren su aplicación de música.

## ¿Qué datos genera?

Spotify es una plataforma de música en línea que tiene más de 600 millones de usuarios en todo el mundo. Cada vez que alguien abre la aplicación, genera una cantidad enorme de información: qué canción escuchó, a qué hora, cuánto tiempo la escuchó, si la saltó o la repitió, qué playlist estaba usando, desde qué país o ciudad se conectó, qué dispositivo usó (celular, computadora, televisor), e incluso si escucha música mientras hace ejercicio o mientras duerme. Todo esto ocurre al mismo tiempo, con millones de usuarios activos simultáneamente, las 24 horas del día. En un solo día, Spotify registra miles de millones de estas acciones.

---

## ¿Quién los usa?

Los datos los recopila y usa principalmente la propia empresa Spotify para mejorar su servicio. También los usan los artistas y sus equipos, quienes pueden ver en tiempo real cuántas personas escuchan sus canciones y desde dónde. Las discográficas los usan para decidir en qué artistas invertir. Los anunciantes los usan para mostrar publicidad personalizada a los usuarios de la versión gratuita. Además, investigadores de mercado y universidades han usado datos de Spotify para estudiar tendencias culturales y comportamiento humano relacionado con la música.

---

## ¿Qué problema resuelve?

El análisis de todos estos datos permite que Spotify sepa exactamente qué tipo de música le gusta a cada persona y en qué momento del día. Con eso, puede crear listas de reproducción personalizadas de forma automática, como el famoso "Descubrimiento Semanal", que le recomienda canciones nuevas a cada usuario basándose en su historial. Esto mejora la experiencia del usuario, hace que la gente use más la aplicación y, en consecuencia, Spotify gana más dinero con suscripciones y publicidad. También ayuda a artistas nuevos a llegar a personas que probablemente les gusten, algo que antes era casi imposible sin una gran discográfica detrás.

---

## ¿Por qué no bastaría Excel?

Excel simplemente no podría con esto. Solo en un día, Spotify registra miles de millones de acciones de usuarios, un número que ni siquiera cabe en una hoja de cálculo, ya que Excel tiene un límite de poco más de un millón de filas. Además, esos datos llegan en tiempo real, segundo a segundo, desde todos los países del mundo al mismo tiempo, algo que Excel no puede procesar mientras ocurre. Los datos tampoco son solo números en una tabla: incluyen texto, audio, imágenes de portadas, ubicaciones geográficas y patrones de comportamiento, tipos de información que no encajan en columnas y filas tradicionales. Por último, para que la recomendación de una canción llegue al usuario en segundos, el sistema necesita analizar millones de datos al instante, y una base de datos relacional común se quedaría paralizada ante esa velocidad.

---

## Conclusiones

1. **El Big Data va más allá del volumen.** El caso de Spotify demuestra que no se trata solo de tener muchos datos, sino de procesarlos a gran velocidad, en tiempo real y en una variedad de formatos que ninguna hoja de cálculo podría manejar. Es la combinación de las 4 Vs lo que convierte a Spotify en un caso real de Big Data.

2. **Los datos generan valor cuando se transforman en decisiones.** Spotify no recopila información solo por acumularla, sino para convertirla en recomendaciones personalizadas, mejores experiencias para el usuario y mejores oportunidades para los artistas. Esto refleja la idea central del curso: el objetivo del Big Data es transformar datos en conocimiento útil.

3. **Las herramientas deben estar a la altura del problema.** Intentar gestionar los datos de Spotify con Excel o una base de datos relacional sería como intentar transportar mil personas en un auto particular. El problema no es la herramienta en sí, sino que no fue diseñada para esa escala. Reconocer cuándo un problema supera las herramientas convencionales es una habilidad fundamental en el área de datos.
